```python
import os
import sys
import pandas as pd
import numpy as np

sys.path.append(os.path.abspath(".."))

# Load files
df1 = pd.read_csv('../data/preds_all.csv')
df2 = pd.read_csv('../data/preds_all_i.csv')

print("===== BASIC INFO =====")
print("df1 shape:", df1.shape)
print("df2 shape:", df2.shape)

print("\n===== COLUMN CHECK =====")
print("Same columns:", set(df1.columns) == set(df2.columns))

print("\nSame column order:")
print(list(df1.columns) == list(df2.columns))

# Show differing column order if needed
if list(df1.columns) != list(df2.columns):
    print("\nColumns with different positions:")
    
    for i, (a, b) in enumerate(zip(df1.columns, df2.columns)):
        if a != b:
            print(f"Position {i}: {a} != {b}")

print("\n===== NULL CHECK =====")
print("Null counts equal:")
print(df1.isnull().sum().equals(df2.isnull().sum()))

print("\n===== EXACT MATCH CHECK =====")
exact_match = df1.equals(df2)
print("Exact dataframe match:", exact_match)

print("\n===== NUMERIC COMPARISON =====")

# Compare numeric columns only
num1 = df1.select_dtypes(include=[np.number])
num2 = df2.select_dtypes(include=[np.number])

# Check approximate equality
all_close = np.allclose(
    num1.values,
    num2.values,
    equal_nan=True
)

print("Numerically close:", all_close)

# Maximum numeric difference
max_diff = np.nanmax(np.abs(num1.values - num2.values))

print("Maximum absolute difference:", max_diff)

print("\n===== DIFFERENCE SUMMARY =====")

# Find cells that differ
diff_mask = ~(num1 == num2)

# Ignore NaN vs NaN
diff_mask = diff_mask & ~(num1.isna() & num2.isna())

num_diffs = diff_mask.sum().sum()

print("Number of differing numeric cells:", num_diffs)

if num_diffs > 0:
    print("\nFirst 10 differences:\n")
    
    diff_locations = np.where(diff_mask)

    for row, col in list(zip(*diff_locations))[:10]:
        col_name = num1.columns[col]
        
        val1 = num1.iloc[row, col]
        val2 = num2.iloc[row, col]

        print(
            f"Row {row}, Column '{col_name}': "
            f"{val1} vs {val2}"
        )
```
